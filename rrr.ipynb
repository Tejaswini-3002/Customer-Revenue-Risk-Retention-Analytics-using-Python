```python
#1 order_value – calculated using quantity × price (function mandatory)
def calculate_order_value(row):
    return row["quantity"] * row["price"]

order_items["order_value"] = order_items.apply(calculate_order_value, axis=1)
```

```python
#5 customer_value_segment – High / Medium / Low using if–else logic
def customer_value_segment(value):
    if value > 5000:
        return "High"
    elif value >= 2000:
        return "Medium"
    else:
        return "Low"
```

```python
def customer_risk_category(rate):
    if rate > 0.30:
        return "High Risk"
    elif rate >= 0.10:
        return "Medium Risk"
    else:
        return "Low Risk"

orders["customer_risk_category"] = orders["return_rate"].apply(customer_risk_category)
```

```python
#7 order_reliability_score – Delivered (+1), Returned (−1), Cancelled (−2)
def order_reliability_score(status):
    if status == "Delivered":
        return 1
    elif status == "Returned":
        return -1
    elif status == "Cancelled": 
        return -2
    else:
        return 0
```

```python
orders["order_reliability_score"] = orders["order_status"].apply(order_reliability_score)
```

```python
#Business Questions to Answer
#1.Which customers generate high revenue but have high return rates?
customer_summary = orders.groupby("customer_id").agg(
    total_revenue=("order_value", "sum"),
    total_orders=("order_id", "nunique"),
    total_returns=("is_returned", "sum")
).reset_index()
customer_summary
```

```python
customer_summary["return_rate"] = (
    customer_summary["total_returns"] / customer_summary["total_orders"])
customer_summary
```

```python
customer_summary["total_revenue"].mean()
```

```python
 customer_summary["return_rate"].mean()
```

```python
high_value_high_return = customer_summary[
    (customer_summary["total_revenue"] > customer_summary["total_revenue"].mean()) &
    (customer_summary["return_rate"] > customer_summary["return_rate"].mean())
]
high_value_high_return = high_value_high_return.sort_values(
    "total_revenue", ascending=False)

high_value_high_return
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(high_value_high_return["customer_id"].astype(str),
        high_value_high_return["total_revenue"])
plt.title("High Revenue Customers with High Return Rates")
plt.xlabel("Customer ID")
plt.ylabel("Total Revenue")
plt.xticks(rotation=90)
plt.show()
```

```python
orders = orders.merge(
    customers[["customer_id", "city"]],
    on="customer_id",
    how="left"
)
orders
```

```python
#2.Which cities suffer most from delivery delays and revenue loss?
orders["revenue_loss"] = orders.apply(
    lambda row: row["order_value"] if row["order_status"] == "Returned" else 0,
    axis=1
)

city_summary = orders.groupby("city").agg(
    total_orders=("order_id", "nunique"),
    delayed_orders=("delivery_delay_days", lambda x: (x > 7).sum()),
    revenue_loss=("revenue_loss", "sum")
).reset_index()

city_summary["delay_rate"] = (
    city_summary["delayed_orders"] / city_summary["total_orders"]
)

city_summary = city_summary.sort_values(
    ["delay_rate", "revenue_loss"],
    ascending=False
)

city_summary
```

```python
#3 Are repeat customers actually more valuable?
customer_order_count = orders.groupby("customer_id")["order_id"].transform("nunique")

orders["is_repeat_customer"] = customer_order_count.apply(lambda x: 1 if x > 1 else 0)

customer_order_count.head()
```

```python
orders["is_repeat_customer"].head()
```

```python
repeat_analysis = orders.groupby("is_repeat_customer").agg(
    avg_order_value=("order_value", "mean"),
    total_revenue=("order_value", "sum"),
    total_customers=("customer_id", "nunique")
).reset_index()

repeat_analysis
```

```python
#4 Which categories appear profitable but lose revenue due to returns?
orders = orders.merge(order_items[["order_id", "category"]],on="order_id",how="left")
orders.head()
```

```python
orders["revenue_loss"] = orders.apply(
    lambda row: row["order_value"] if row["order_status"] == "Returned" else 0,axis=1)
orders
```

```python
category_summary = orders.groupby("category").agg(
    total_revenue=("order_value", "sum"),
    revenue_loss=("revenue_loss", "sum")
).reset_index()

category_summary["loss_percentage"] = (
    category_summary["revenue_loss"] / category_summary["total_revenue"]
)

category_summary = category_summary.sort_values(
    "loss_percentage",
    ascending=False
)

category_summary
```

```python
#5 Which payment mode carries the highest risk?
orders["is_returned"] = orders["order_status"].apply(lambda x: 1 if x == "Returned" else 0)
orders.head()
```

```python
payment_summary = orders.groupby("payment_mode").agg(
    total_orders=("order_id", "nunique"),
    total_returns=("is_returned", "sum"),
    total_revenue=("order_value", "sum")
).reset_index()

payment_summary["return_rate"] = (
    payment_summary["total_returns"] / payment_summary["total_orders"]
)

payment_summary = payment_summary.sort_values(
    "return_rate",
    ascending=False
)

payment_summary
```
